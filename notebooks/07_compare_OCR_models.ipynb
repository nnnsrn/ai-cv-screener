```python
import os
import time
import tracemalloc
import pandas as pd
import numpy as np
import gc
from PIL import Image
import pymupdf
```

```python
# --- 1. Setup Directories & Target Files ---
test_dir = "benchmark_test_files"
output_text_dir = "output_ocr_comparison_new"
os.makedirs(output_text_dir, exist_ok=True)

target_files = [f"image_based_{i}.pdf" for i in range(1, 11)]

def save_text_to_file(original_file, library_name, extracted_text):
    if extracted_text:
        safe_name = f"{original_file}_{library_name}.txt"
        out_path = os.path.join(output_text_dir, safe_name)
        with open(out_path, "w", encoding="utf-8") as f:
            f.write(extracted_text)
```

```python
# --- 2. Initialize Models Safely ---
print("⏳ Initializing OCR Engines...")
models_ready = {}
```

```python
# EasyOCR
try:
    import easyocr
    easy_reader = easyocr.Reader(['en'], gpu=False, verbose=False)
    models_ready['EasyOCR'] = True
except Exception as e:
    print(f"EasyOCR Init Warning: {e}")
    models_ready['EasyOCR'] = False
```

```python
# RapidOCR
try:
    from rapidocr_onnxruntime import RapidOCR
    rapid_reader = RapidOCR()
    models_ready['RapidOCR'] = True
except Exception as e:
    print(f"RapidOCR Init Warning: {e}")
    models_ready['RapidOCR'] = False
```

```python
# PaddleOCR
try:
    from paddleocr import PaddleOCR
    paddle_reader = PaddleOCR(use_textline_orientation=False, lang='en', enable_mkldnn=False)
    models_ready['PaddleOCR'] = True
except Exception as e:
    print(f"PaddleOCR Init Warning: {e}")
    models_ready['PaddleOCR'] = False
```

```python
# DocTR
try:
    from doctr.models import ocr_predictor
    doctr_model = ocr_predictor(pretrained=True)
    models_ready['DocTR'] = True
except Exception as e:
    print(f"DocTR Init Warning: {e}")
    models_ready['DocTR'] = False
```

```python
# Surya OCR
try:
    from surya.recognition import RecognitionPredictor
    from surya.detection import DetectionPredictor
    
    surya_det_predictor = DetectionPredictor()
    surya_rec_predictor = RecognitionPredictor()
    models_ready['SuryaOCR'] = True
except Exception as e:
    print(f"Surya OCR Init Warning: {e}")
    models_ready['SuryaOCR'] = False
```

```python
# --- 3. Execution & Measurement Loop ---
results = []
print(f"\n🚀 Starting Benchmark on target files...\n")

for file_name in target_files:
    file_path = os.path.join(test_dir, file_name)
    
    if not os.path.exists(file_path):
        print(f"⚠️ Skipping {file_name}: File not found.")
        continue
        
    print(f"Analyzing: {file_name}...")
    
    doc = pymupdf.open(file_path)
    img_arrays = []
    pil_images = []
    for page in doc:
        pix = page.get_pixmap(dpi=150)
        img = Image.frombytes("RGB", [pix.width, pix.height], pix.samples)
        pil_images.append(img)
        img_arrays.append(np.array(img))
        
    # EasyOCR
    if models_ready.get('EasyOCR'):
        tracemalloc.start()
        t0 = time.time()
        try:
            text = "\n".join([" ".join(easy_reader.readtext(arr, detail=0)) for arr in img_arrays])
            save_text_to_file(file_name, "EasyOCR", text)
        except Exception: text = ""
        current, peak = tracemalloc.get_traced_memory()
        tracemalloc.stop()
        results.append({"File": file_name, "Tool": "EasyOCR", "Time (s)": round(time.time() - t0, 4), "Memory (MB)": round(peak / 10**6, 2), "Chars": len(text.strip())})

    # RapidOCR
    if models_ready.get('RapidOCR'):
        tracemalloc.start()
        t0 = time.time()
        try:
            text = ""
            for arr in img_arrays:
                res, _ = rapid_reader(arr)
                if res: text += "\n".join([line[1] for line in res]) + "\n"
            save_text_to_file(file_name, "RapidOCR", text)
        except Exception: text = ""
        current, peak = tracemalloc.get_traced_memory()
        tracemalloc.stop()
        results.append({"File": file_name, "Tool": "RapidOCR", "Time (s)": round(time.time() - t0, 4), "Memory (MB)": round(peak / 10**6, 2), "Chars": len(text.strip())})

    # PaddleOCR
    if models_ready.get('PaddleOCR'):
        tracemalloc.start()
        t0 = time.time()
        try:
            text = ""
            for arr in img_arrays:
                res = paddle_reader.predict(arr)
                if res and res[0]: text += "\n".join([line[1][0] for line in res[0]]) + "\n"
            save_text_to_file(file_name, "PaddleOCR", text)
        except Exception: text = ""
        current, peak = tracemalloc.get_traced_memory()
        tracemalloc.stop()
        results.append({"File": file_name, "Tool": "PaddleOCR", "Time (s)": round(time.time() - t0, 4), "Memory (MB)": round(peak / 10**6, 2), "Chars": len(text.strip())})

    # DocTR
    if models_ready.get('DocTR'):
        tracemalloc.start()
        t0 = time.time()
        try:
            text = ""
            doc_res = doctr_model(img_arrays)
            for page in doc_res.pages:
                for block in page.blocks:
                    for line in block.lines:
                        text += " ".join([word.value for word in line.words]) + "\n"
            save_text_to_file(file_name, "DocTR", text)
        except Exception: text = ""
        current, peak = tracemalloc.get_traced_memory()
        tracemalloc.stop()
        results.append({"File": file_name, "Tool": "DocTR", "Time (s)": round(time.time() - t0, 4), "Memory (MB)": round(peak / 10**6, 2), "Chars": len(text.strip())})

    # Surya OCR
    if models_ready.get('SuryaOCR'):
        tracemalloc.start()
        t0 = time.time()
        try:
            text = ""
            # Surya takes PIL images directly
            predictions = surya_rec_predictor(pil_images, det_predictor=surya_det_predictor)
            for page_pred in predictions:
                text += "\n".join([line.text for line in page_pred.text_lines]) + "\n"
            save_text_to_file(file_name, "SuryaOCR", text)
        except Exception as e:
            text = ""
        current, peak = tracemalloc.get_traced_memory()
        tracemalloc.stop()
        results.append({
            "File": file_name, 
            "Tool": "SuryaOCR", 
            "Time (s)": round(time.time() - t0, 4), 
            "Memory (MB)": round(peak / 10**6, 2), 
            "Chars": len(text.strip())
        })
        
    gc.collect()
```

```python
# --- 4. Final Output ---
df_resources = pd.DataFrame(results)
print("\n OCR BENCHMARK COMPLETE!")
try:
    from IPython.display import display
    display(df_resources)
except ImportError:
    print(df_resources.to_string())
```

```python
import os
from jiwer import cer, wer

output_dir = "output_ocr_comparison_new"
target_files = [f"image_based_{i}.pdf" for i in range(1, 11)]

baseline_tool = "DocTR"
comparison_tools = ["EasyOCR", "RapidOCR", "PaddleOCR", "SuryaOCR"]

def load_text(filename):
    path = os.path.join(output_dir, filename)
    if os.path.exists(path):
        with open(path, "r", encoding="utf-8") as f:
            return f.read().strip()
    return ""

print(f" Relative Agreement Score against (Baseline: {baseline_tool})")
print("=" * 60)

for file_name in target_files:
    baseline_text = load_text(f"{file_name}_{baseline_tool}.txt")
    
    if not baseline_text:
        continue
        
    print(f"\n📄 {file_name}:")
    
    for tool in comparison_tools:
        tool_text = load_text(f"{file_name}_{tool}.txt")
        if tool_text:
            if len(tool_text) == 0:
                print(f"    {tool}: FAILED (0 characters)")
                continue
                
            char_error = cer(baseline_text, tool_text)
            word_error = wer(baseline_text, tool_text)
            
            char_accuracy = max(0, (1 - char_error) * 100)
            word_accuracy = max(0, (1 - word_error) * 100)
            
            print(f"   {tool}: Char Acc: {char_accuracy:.1f}% | Word Acc: {word_accuracy:.1f}%")
```

Rumus tersebut menggunakan Levenshtein Distance (jarak edit minimum) yang dihitung oleh library jiwer. Metrik dasarnya adalah CER (Character Error Rate) dan WER (Word Error Rate).
S (Substitutions): Jumlah karakter atau kata yang salah dibaca oleh engine (contoh: baseline "Python", OCR membaca "Pyth0n" => S=1).
D (Deletions): Jumlah karakter atau kata pada dokumen asli yang terlewat atau gagal diekstrak (contoh: baseline "Machine Learning", OCR membaca "Machine" => D=1).
I (Insertions): Jumlah karakter atau kata artefak/halusinasi yang muncul pada hasil OCR namun tidak ada di referensi (contoh: baseline "Data", OCR membaca "Data ||" => I=2).
N: Total panjang karakter atau kata pada teks acuan (baseline_text).

Analisis Bias pada Penggunaan Baseline 
Penggunaan model (seperti DocTR) sebagai referensi tanpa verifikasi manual manusia menimbulkan keterbatasan:
a) Potensi Bias:Ketika menetapkan engine tertentu sebagai pembanding utama (pseudo-ground truth), pengujian ini secara teknis mengukur Tingkat Keselarasan/Konsistensi Antar-Model (Inter-model Agreement atau Relative Similarity), alih-alih mengukur Ground Truth Accuracy.
b) Layout Misinterpretation: Jika model baseline membaca dokumen berformat dua kolom secara keliru dari atas ke bawah secara vertikal, sementara engine lain membaca secara horizontal sesuai alur normal manusia, engine uji akan mendapatkan penalti skor akurasi yang buruk meskipun susunan pembacaannya sebenarnya lebih akurat.
c) Insertion Penalties: Jika model baseline gagal mendeteksi bagian teks berukuran kecil di area header atau footer dokumen, namun engine uji berhasil menemukannya, teks tambahan tersebut justru dikalkulasikan oleh jiwer sebagai bentuk kesalahan Insertion atau halusinasi, sehingga menjatuhkan skor akurasi model uji.
